```python
import requests
from bs4 import BeautifulSoup
from langchain_openai import AzureChatOpenAI


#Extração de dados da URL selecionada
def extract_text_from_url(url):
    response = requests.get(url)

    if response.status_code == 200:
        soup = BeautifulSoup(response.text, 'html.parser')
        for script_or_style in soup(['script', 'style']):
            script_or_style.decompose()
        texto = soup.get_text(separator=' ')
        linhas = (line.strip() for line in texto.splitlines())
        texto_limpo = '\n'.join(line for line in linhas if line)
        return texto_limpo
    else:
        print(f"Failed to fetch the URL. Status code: {response.status_code}")
        return None

#Conexão com os dados da Azure
#Substituir com os seus dados:
client = AzureChatOpenAI(
    azure_endpoint = "seu_endpoint,
    api_key = "chave_da_api",
    api_version = "versao_da_api_em_dd/mm/aaaa",
    deployment_name = "nome_da_ia_usadea",
    max_retries = numero_de_tentativas
)

#Chamada da função
def translate_article(text, lang):
    messages = [
        ("system" , "Você atua como tradutor de textos"),
        ("user" , f"Traduza o {text} para o idioma {lang} e responda em markdown")
    ]

    response = client.invoke(messages)
    print(response.content)
    return response.content
```

Agora vamos realizar a tradução!
Insira a URL do artigo que você deseja traduzir e execute o código!
Após isso o artigo escolhido irá ser apresentado traduzido abaixo.

```python
url = "url_do_artigo"
texto_extraido = extract_text_from_url(url)
article = translate_article(texto_extraido , "pt-br")
print(article)
```
